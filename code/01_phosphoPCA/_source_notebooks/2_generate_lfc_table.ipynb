### blib_fp
1. The count table was generated using an empty subtrate library file, and so should be further sanitized

```python
DATADIR = "${KSI_DATA_ROOT}/sequencing_data_1217/02_fastp_out/"
filename = 'blib_fp.count.txt'
dat = pd.read_csv(os.path.join(DATADIR,filename),sep = '\t')
```

```python
dat_filtered = dat.loc[dat['sgRNA'].apply(lambda s: len(s.split('GGGAGGTGGAGCTAGC'))>1),:]
```

```python
dat_filtered = dat_filtered.loc[dat_filtered['sgRNA'].apply(lambda s: len(s.split('TGGATCAGGTGGTTAGAAGCTTATTAGTT'))>1),:]
```

```python
dat_filtered['sgRNA_trimmed'] = dat_filtered['sgRNA'].apply(lambda s: s.split('GGGAGGTGGAGCTAGC')[1].split('GGTGGATCAGGTGGTTAGAAGCTTATTAGTT')[0])
```

```python
dat_filtered.loc[dat_filtered['sgRNA_trimmed'].apply(lambda s: len(s)==45)].shape[0]/dat_filtered.shape[0]
```

```python
dat_filtered = dat_filtered.loc[dat_filtered['sgRNA_trimmed'].apply(lambda s: len(s)==45)]
dat_filtered['Gene'] = dat_filtered['sgRNA_trimmed'].apply(lambda s: str(Seq(s).translate()))
```

```python
dat_filtered.head(5)
```

```python
dat_filtered.to_csv('${KSI_DATA_ROOT}/processed_count_table/blib_fp.count.txt',sep = '\t',index = False)
```

```python
dat_filtered = pd.read_csv('${KSI_DATA_ROOT}/processed_count_table/blib_fp.count.txt',sep = '\t')
dat_filtered_sub = dat_filtered.loc[:,['sgRNA','Gene','blib10_pre','blib10_mtx']]
dat_filtered_sub = dat_filtered_sub.loc[dat_filtered_sub.iloc[:,2]+dat_filtered_sub.iloc[:,3]>100,:]
print(dat_filtered_sub.shape) 
dat_filtered_sub.to_csv('${KSI_DATA_ROOT}/processed_count_table/blib_fpblib10.count.txt',sep = '\t',index = False)
```
