```python
import pandas as pd
```

```python
human_yeast_pos = pd.read_pickle("../../data/training_data/mif_training_230628/test_humanyeast.pkl")
```

```python
all_neg = pd.read_pickle("../../data/training_data/mif_training_230628/train_nodark.pkl")
```

```python
all_neg = all_neg.loc[all_neg["label"]==0,:]
```

```python
yeast_all_neg = all_neg.loc[all_neg["kinase"].isin(human_yeast_pos["kinase"].drop_duplicates()),:]
```

```python
yeast_all_neg.groupby("kinase").count()
```

```python
neg_rows = []
for i in range(human_yeast_pos.shape[0]):
    curr_row = human_yeast_pos.iloc[i,:]
    curr_kin = curr_row["kinase"]
    print(curr_kin)
    try:
        neg_sel = all_neg.loc[all_neg["kinase"]==curr_kin,:].sample(1)
        neg_rows.append(neg_sel)
    except:
        continue
```

```python
yeast_neg_df = pd.concat(neg_rows,axis = 0)
```

```python
yeast_1n1p = pd.concat([human_yeast_pos,yeast_neg_df],axis = 0)
```

```python
yeast_1n1p.groupby("label").count()
```

```python
yeast_1n1p.to_pickle("../../data/benchmark_sets/yeast_human_1n1p.pkl")
```
